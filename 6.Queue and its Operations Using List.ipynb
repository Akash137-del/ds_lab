```python
# EXPERIMENT 6
# Queue Implementation

class Queue:

    def __init__(self):
        self.queue = []

    # Insert element
    def EnQueue(self, item):
        self.queue.append(item)

    # Delete element
    def DeQueue(self):
        if len(self.queue) == 0:
            print("Queue Underflow")
            return None

        return self.queue.pop(0)

    # Display queue
    def display(self):
        if len(self.queue) == 0:
            print("Queue is Empty")
        else:
            print("Queue:", self.queue)

    # Peek front element
    def peek(self):
        if len(self.queue) == 0:
            return None
        return self.queue[0]


# Create Queue
q = Queue()

q.EnQueue(10)
q.EnQueue(20)

q.display()

print("Dequeued item:", q.DeQueue())
print("Dequeued item:", q.DeQueue())

q.EnQueue(30)
q.EnQueue(40)
q.EnQueue(50)

q.display()

print("Dequeued item:", q.DeQueue())

q.display()
```
