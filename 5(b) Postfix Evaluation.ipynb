```python
# EXPERIMENT 5(b)
# Postfix Expression Evaluation

class Evaluate:
    def __init__(self):
        self.stack = []

    def isEmpty(self):
        return len(self.stack) == 0

    def push(self, value):
        self.stack.append(value)

    def pop(self):
        if not self.isEmpty():
            return self.stack.pop()
        return None

    def evaluatePostfix(self, exp):

        for ch in exp:

            if ch.isdigit():
                self.push(int(ch))

            else:
                val1 = self.pop()
                val2 = self.pop()

                if ch == '+':
                    result = val2 + val1
                elif ch == '-':
                    result = val2 - val1
                elif ch == '*':
                    result = val2 * val1
                elif ch == '/':
                    result = val2 / val1
                elif ch == '^':
                    result = val2 ** val1

                self.push(result)

        return self.pop()


# Postfix expression
exp = "231*+9-"

obj = Evaluate()

print("Postfix Expression:", exp)
print("Postfix Evaluation:", obj.evaluatePostfix(exp))
```
